## Dictionaries

```python
config = {
    "browser" : "Brave",
    "aut" : "jwt",
    "role" : "admin",
    "status" : "active"
}

print("Browser ", config.get("browser"))
print("Status ",config.get("status"))

configs = [config]
print(configs)
print(type(configs))
```

# Functions

```python
def getConfigValue(key):
    return config.get(key)

print(getConfigValue("role"))
```

# Class

```python
class UserLogin:

    def __init__(self, secret):
        self.secretPass = secret

    def login(self, username, password):
        if(self.secretPass == password):
            return "Logged In"
        else:
            return "Wrong Password!"

usrlogin = UserLogin("12345")

print(usrlogin.login("ishan", "12345"))
```
